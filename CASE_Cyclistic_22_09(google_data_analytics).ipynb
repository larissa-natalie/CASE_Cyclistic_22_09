1º LEITURA DOS ARQUIVOS USANDO PANDAS

```python
import pandas as pd

df_registro_2019 = pd.read_csv("/content/registros_viagens_2019.csv")
display(df_registro_2019.head()) #mostrar as primeiras linhas da base

print('\n\n\n') #pular linha entre os resultados

df_registro_2020 = pd.read_csv("/content/registros_viagens_2020.csv")
display(df_registro_2020.head())
```

```python
# BASE 2019

#mostra a quantidade de linhas no total, nomes das colunas existentes, nºs de linhas por coluna e tipo de dado (se é nº inteiro-int, decimal-float ou texto-object)
print(df_registro_2019.info())

print('\n') #pula linha entre os resultados

#mostra estatísticas resumidas - média, contagem, valor mínimo, valor máximo - por coluna quantitativa
display(df_registro_2019.describe())

print('\n')

#estatiticas das colunas qualitativas (texto)
display(df_registro_2019.describe(include=object))

print('\n')

#verificando se existem dados duplicados
print(df_registro_2019.duplicated().sum())
```

3º ARRUMANDO OS DADOS PARA DEPOIS REALIZAR A JUNÇÃO DAS 2 BASES

```python
# 1) alterando dados da coluna usertype de 2019, pois não condizem com os dados da base 202
# então vamos mudar "Subscriber" para "member" e "Customer" para "casual"

#usando dicionario para mudar os 2 tipos de 1x
df_registro_2019['usertype'] = df_registro_2019['usertype'].replace({
    'Subscriber': 'member',
    'Customer': 'casual'
})
display(df_registro_2019.head())
```

```python
# 2) arrumando os nomes das colunas da base de 2019 para ficarem iguais as da base de 2020
# isso vai facilitar muito no momento da junção, pois a correspondencia é pelo nome da coluna

df_registro_2019 = df_registro_2019.rename(columns={
    'trip_id': 'ride_id',
    'start_time': 'started_at',
    'end_time': 'ended_at',
    'from_station_id': 'start_station_id',
    'from_station_name': 'start_station_name',
    'to_station_id': 'end_station_id',
    'to_station_name': 'end_station_name',
    'usertype': 'member_casual'
})
display(df_registro_2019.head(1))
```

4º JUNÇÃO DAS BASES

```python
# como queremos juntas duas bases de anos ==, mas com os mesmos nomes de colunas, usamos o metodo .concat (vai juntar as duas em uma base unica = aumentando o historico de registros)

# 1. Definimos quais colunas queremos manter em ambos os datasets (apenas as que realmente fazem sentido para a pergunta de negocio inicial)
# usando um dicionario para facilitar a menção no código
colunas = [
    'ride_id',
    'started_at',
    'ended_at',
    'start_station_name',
    'end_station_name',
    'member_casual'
]

# 2. usar .concat para juntar (usam o termo empilhar)
df_registro_unificado = pd.concat([df_registro_2019[colunas], df_registro_2020[colunas]], ignore_index=True) # ignore_index=True serve para renumerar as linhas do começo ao fim quando você junta duas ou mais tabelas.

# verificando a quantidade de linhas e colunas (precisa ser a soma exata da base 2019+2020)
print(df_registro_unificado.shape)
```

5º ALTERAÇÃO DO TIPO DE DADO DA DATA (object para datatime)

```python
# no .info() as colunas de data 'started_at' e ''ended_at' estão como object,
# isso não é algo bom, o correto é elas serem datatime, então precisamos realizar essa alteração
# sintaxe:  df['nome_da_coluna'] = pd.to_datetime(df['nome_da_coluna'])

df_registro_unificado['started_at'] = pd.to_datetime(df_registro_unificado['started_at'])
df_registro_unificado['ended_at'] = pd.to_datetime(df_registro_unificado['ended_at'])

print(df_registro_unificado.info())
```

6° CRIANDO COLUNAS

```python
# 1° criar uma coluna que mostre a duração da viagem 'ride_length'
# só usar as colunas started_at e ended_at, fznd a subtração

# Criando a coluna 'ride_length' e atribuindo o resultado do cálculo a ela
# nova_variavel['nome_nova_coluna'] = oq ela deve receber - nesse caso o calculo da subtração do tempo
# o .dt.total_seconds() converte a diferenã para o total de segundos
# e a divisão por 60 é para obter a duração em MINUTOS DECIMAIS (ex.: 2.5 minutos = 2 min e 30 seg), para facilitar para fzr media e afins
df_registro_unificado['ride_length'] = ((df_registro_unificado['ended_at'] - df_registro_unificado['started_at']).dt.total_seconds() / 60).round(2)


# 2° coluna que mostra o dia da semana em que a viagem começou 'day_of_week'
# nova_variavel['nome_nova_coluna'] = oq ela deve receber - nesse caso quero q ela pegue o dia da coluna started_at
df_registro_unificado['day_of_week'] = df_registro_unificado['started_at'].dt.day_name()

# conferindo se deu certo
display(df_registro_unificado.head(5))
```

7° FILTRAGEM ANTES DA ANÁLISE

```python
# tirando possiveis viagens com duração negativa ou igual a zero
# (por exemplo, quando a bicicleta é retirada para manutenção ou quando ocorre uma falha no sistema de travamento da estação).
# entro filtro pela coluna de duração da viagem tudo que for >0 ---- essa é uma etapa apenas por segurança
df_limpo = df_registro_unificado[df_registro_unificado['ride_length'] > 0]
```

8º ANÁLISES

```python
# 1° Pergunta: Qual é a média de tempo de viagem (ride_length) para cada tipo de cliente (member vs casual)?
# primeiro agrupa por tipo de cliente e dps faz a média
media_por_usuario = df_limpo.groupby('member_casual')['ride_length'].mean()
display(media_por_usuario)
```

```python
# 2° Pergunta: Qual é o dia da semana (day_of_week) mais popular para cada tipo de usuário?
dia_popular_por_usuario = df_limpo.groupby(['member_casual', 'day_of_week'])['ride_id'].count().sort_values(ascending=False)
display(dia_popular_por_usuario)
```

```python
# 3° Pergunta: Em que horas do dia os membros e os casuais mais utilizam as bicicletas?

# primeiro vms criar a coluna para hrs, extraindo do inicio da viagem
df_limpo['hour'] = df_limpo['started_at'].dt.hour

# segundo fazemos a conta
horario_pico = df_limpo.groupby(['member_casual','hour'])['ride_id'].count()
display(horario_pico)
```

```python
from numpy import frexp
# 4° Pergunta: Quais os locais mais frequentados pelos clientes casuais?
# Isso ajudará o mkt para saber ond fzr publis ou ativações

frenquencia_casuais = df_limpo[df_limpo['member_casual'] == 'casual']['start_station_name'].value_counts()
display(frenquencia_casuais.head(3))
```

8° GRÁFICOS

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Definimos e aplicamos a ordem correta dos dias da semana
ordem_dias = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df_limpo['day_of_week'] = pd.Categorical(df_limpo['day_of_week'], categories=ordem_dias, ordered=True)

# 2. Criamos a figura com um tamanho adequado
plt.figure(figsize=(12, 6))

# 3. Geramos o gráfico de barras
ax = sns.countplot(
    data=df_limpo,
    x="day_of_week",
    hue="member_casual",
    palette="Set2"
)

# 4. Adicionamos os números no topo de cada barra 🔢
for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', padding=3, fontsize=9)

# 5. Personalizamos os títulos e eixos
plt.title("Total de Viagens por Dia da Semana e Tipo de Utilizador 🚴", fontsize=14, fontweight='bold')
plt.xlabel("Dia da Semana", fontsize=12)
plt.ylabel("Quantidade de Viagens", fontsize=12)
plt.legend(title="Tipo de Utilizador")

# 6. Exibimos o gráfico limpo
plt.show()
```

```python
# 1. Contagem dos dados para o gráfico de pizza
# O erro ocorre porque qtd_clientes é um único número (o total de linhas),
# mas plt.pie espera uma sequência de valores para as fatias do gráfico.
# Precisamos contar quantos 'member' e quantos 'casual' existem.
proporcao_clientes = df_limpo['member_casual'].value_counts()

# 2. Criar a figura
plt.figure(figsize=(6, 6))

# 3. Desenhar o gráfico de pizza limpo
plt.pie(
    proporcao_clientes, # Agora passamos a série com as contagens de cada tipo
    labels=['Membros', 'Casuais'], # Os rótulos devem corresponder à ordem dos valores em proporcao_clientes
    autopct='%1.1f%%',  # Exibe a percentagem 🔢
    colors=['#2b5c8f', '#d95f02'],
    startangle=90
)

# 4. Adicionar título
plt.title('Proporção de Utilizadores: Membros vs. Casuais')

# 5. Exibir
plt.show()
```
